```python

import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Company": ["Apple", "Google", "Microsoft", "Amazon"],
    "Stock Price": [250, 180, 420, 220]
}

df = pd.DataFrame(data)

plt.figure(figsize=(8, 5))
plt.bar(df["Company"], df["Stock Price"])

plt.title("Stock Market Dashboard")
plt.xlabel("Company")
plt.ylabel("Sample Stock Price ($)")

plt.show()
```

```python

import pandas as pd

data = {
    "Company": ["Apple", "Google", "Microsoft", "Amazon"],
    "Stock Price": [250, 180, 420, 220]
}

df = pd.DataFrame(data)

company = input("Enter company name (Apple, Google, Microsoft, Amazon): ")

result = df[df["Company"].str.lower() == company.lower()]

if not result.empty:
    print("Company:", result.iloc[0]["Company"])
    print("Sample Stock Price: $", result.iloc[0]["Stock Price"])
else:
    print("Company not found. Please try again.")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
plt.plot(df["Company"], df["Stock Price"], marker="o")
plt.title("Stock Market Line Chart")
plt.xlabel("Company")
plt.ylabel("Sample Stock Price ($)")
plt.grid(True)
plt.show()
```

```python
highest = df.loc[df["Stock Price"].idxmax()]

print("Company with highest stock price:", highest["Company"])
print("Highest stock price:", highest["Stock Price"])
```

```python
print(df[["Company", "Stock Price"]])
```

```python
average = df["Stock Price"].mean()

print("Average Stock Price:", average)
```

```python
lowest = df.loc[df["Stock Price"].idxmin()]

print("Company with lowest stock price:", lowest["Company"])
print("Lowest stock price:", lowest["Stock Price"])
```

```python
sorted_df = df.sort_values(by="Stock Price")

print(sorted_df)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
plt.barh(df["Company"], df["Stock Price"])
plt.title("Company Stock Price Comparison")
plt.xlabel("Stock Price ($)")
plt.ylabel("Company")
plt.show()
```
